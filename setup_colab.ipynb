## 1단계 — Webhook URL을 Secrets에 저장

1. 좌측 사이드바의 🔑 (열쇠) 아이콘 클릭 → **새 보안 비밀** 추가
2. **이름**: `SLACK_WEBHOOK_URL`
3. **값**: 친구한테 받은 URL (`https://hooks.slack.com/services/...` 형태)
4. **노트북 액세스** 토글 → ON

끝났으면 아래 셀 실행.
